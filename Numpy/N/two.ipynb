```python
import numpy as np
```

```python
#mathemetical operations on arrays in python
arr = np.array([10,20,30,40,50]) 
print(arr * 2)
print(arr + 5)
print(arr ** 2)
```

```python
#aggregation functions
print(np.sum(arr))
print(np.min(arr))
print(np.max(arr))
print(np.std(arr))
print(np.var(arr))
print(np.mean(arr))
```
