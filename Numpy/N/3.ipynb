```python
import numpy as np
```

```python
arr = np.array([1,2,3,4,5,6])
print(arr)
```

```python
new_arr = np.insert(arr,2,100)
print(new_arr)
```

```python
#2d array
arc = np.array([[1,2],[5,6]])
# print(arc)
new_arc = np.insert(arc,1,[3,4],axis=0) #we adding a new element at index 1 axis 0 wil add to the rows and axis 1 wil add it to columns 
print(new_arc)
```

```python
m = np.array([10,20,30])
new_ar = np.append(m,[40,50,60]) #append element to last index
print(new_ar)
```

```python
#concate the arrays
r = np.concatenate(([[[22,9],[8,65]]]),axis=1)
print(r)
```

```python
#delete function to delete specefic elements at a index
c = np.delete(r,2)
print(c)
```

```python
rt= np.delete(r,0,axis=0) #in the upper cell there was no row so we did not give any axis but in this one it 2d array so we need to give a row that is 0 
print(rt)
```

```python
#staking and splting
ard = np.array([1,2,3])
ard1 = np.array([1,23,5])
print(np.vstack((ard,ard1))) #vertically stacking the arrays
c = np.hstack((ard,ard1)) #horizontaliy stacking the arrays
print(c)
```

```python
#splting the in parts of the sub arrays
print(np.split(arr,2))
```

```python
#matchemetical operations brodcasting
prices = np.array([100,500,400])
discount = 10

final_price = prices - (prices * discount/100)
print(final_price)

#single value brodcasting
result = prices * 2
print(result)

#2d array brodcasting
app = np.array([[12,3,5],[4,5,9]])
vector = np.array([15,8,4])

kk = app + vector
print(kk)
```

```python
#error will be there because we cant add any vlaue that does not match the shape 
arrrr = np.array([[12,3,5],[4,5,9]])
th = np.array([1,2]).reshape(2,1) # here we reshaped the 1d array to 2d with 2 rows and 1 comumns so we can methemetical operations
mh = arrrr + th
print(mh)
```

```python
#vectoriztion  faster then pythin loops
mb1 = np.array([45,9,8])
mb2 = np.array([14,5,7])
ccccc = mb1 + mb2
print(ccccc)
```

```python
#handling missing value
#first cheking that if we have not a number value nan
anan = np.array([1,2,3,np.nan,5,8,np.nan])
anac = np.isnan(anan)
print(anac)

#replacing the nan vlaue to number default is 0
clean = np.nan_to_num(anan,nan=100)
print(clean)
```

```python
#handling infinte value cheaking it

anat = np.array([1,2,3,np.inf,5,8,-np.inf])
c = np.isinf(anat)
print(c)

#cleaning the value 
cleaned_ = np.nan_to_num(anat,posinf=1000,neginf=-1000)
print(cleaned_)
```
