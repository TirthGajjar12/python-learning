```python
#cheaking the arrays shape size and type 
#shape
arr_2d = np.array([[[1,2],[4,5],[7,8.7]]])
print(arr_2d.shape) #it shows 2 rows and 3 rows it retuns tuple
print(arr_2d.size) #it shows size there are 6 elements in the matrics or 2d array
print(arr_2d.ndim) #number os dimansions here we have 2d arrays so it shows 2
print(arr_2d.dtype) #checking of the datatype in the arrays


```

```python
#indexing and slicing in numpy
krr = np.array([10,20,30,40,50])
print(krr[0])
print(krr[2])
print(krr[-1])
```

```python
#slicing arrays
print(krr[::-1]) #reversing the array
print(krr[1:3])
print(krr[::2])
```

```python
#fancy inexing means selecting multiple elements at onces
print(krr[[0,1,4]])
```

```python
#filtering data // boolean masking
print(krr[krr>20])
```

```python
#resaping and manupulating the arrays 
mrr = np.array([1,2,3,4,5,6])
reshaped = mrr.reshape(2,3) #first 2 here is row and 3 is colum
print(reshaped) #it does not return a copy or create copy it returns view measn original areay will be modified
```

```python
#flating arrays
print(mrr.ravel()) #revel will return view means original with modification
print(mrr.flatten())#flatten will retrn a copy and will not modify the orginal
```
