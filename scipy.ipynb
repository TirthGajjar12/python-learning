```python
from scipy import stats
import numpy as np
# print(constants.pi)
```

```python
#unit categories : matrix , bunary ,mass,angle,time,length,pressure,volume,speed,temperature,engery,power

# print(constants.yocto)

response_times = [100, 110, 110, 120, 500]

mean = np.mean(response_times)
print(mean)
```

```python
#spance data have zeros and dense data have no zeros in them

median = np.median(response_times)
print(median)
```

```python
mode = stats.mode(response_times,keepdims=False)
print(mode.mode)
```

```python
#tandard deviation (STD) — How spread out is the data
#A larger standard deviation means more variation in your measurements. A smaller one means the measurements are more consistent

A = [100, 105, 100, 95, 100]
B = [50, 150, 80, 120, 100]

Astd = stats.tstd(A)
print(Astd)
Bstd = stats.tstd(B)
print(Bstd)
```

```python
#Skewness tells you whether the data is leaning towards one side.
#Imagine: 10, 20, 20, 25, 30, 200 Most values are small, but one value is very large.

skewness = stats.skew(response_times,bias=False)
print(skewness)
```

```python
#Kurtosis helps describe how likely a distribution is to contain extreme values relative to a normal distribution
#Website A usually responds between 90 and 110 ms.
#Website B usually responds between 90 and 110 ms, but occasionally takes 2,000 ms.

kurt = stats.kurtosis(response_times,fisher=True,bias=False)
print(kurt)
```

```python
#probability distribution
#normal distribution
#loc=100 means average.
#scale=10 means standard deviation.
#sf(120) calculates the probability of being above 120.
prob = stats.norm.sf(120,loc=100,scale=10)
print(prob) #its is 2.275%
```

```python
#binomial distribtion
#unserstanding the perametrs 2 means exact 2 faliures , n=10 means total attampts and p=0.10 means 10 percentage chances of failure
probb = stats.binom.pmf(2,n=10,p=0.10)
print(probb) #its 19.37%
```

```python
#Poisson distribution
#mu=3 means the average number of events in the chosen interval is 3.
probp  = stats.poisson.pmf(5,mu=3)
print(probp) #it is 10%
```

#when which is used

Normal = Bell-shaped measurements.



Binomial = Number of successes in a fixed number of attempts.



Poisson = Number of events in a fixed interval.


```python
#4. Hypothesis testing
version_a = [120, 115, 130, 125, 118]

version_b = [140, 135, 150, 145, 138]

result = stats.ttest_ind(version_a,version_b,equal_var=False)

print("T-statistic:", result.statistic)
print("P-value:", result.pvalue)
```


- Version A and B are two independent groups.
- The t-test compares their average response times.
- equal_var=False uses Welch's t-test, which does not assume the groups have equal variances.
There are other types too:
Test	When to use
One-sample t-test	Compare one group's average against a known value
Independent t-test	Compare two separate groups
Paired t-test	Compare the same subjects before and after a change
