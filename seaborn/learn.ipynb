```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly as px
```

```python
tips = sns.load_dataset('tips')
tips
```

```python
#scatter plot axis level fucntion
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
#figure level function
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',style='time',size='size')
```

```python
#line polot we use this when we are dealing with time duration
gap =px.data.gapminder()
gap
```

```python
# temp_tm = gap[gap['country'] == 'India' ]
temp_df = gap[gap['country'].isin(['India', 'China', 'Pakistan'])]
temp_df
```

```python
temp_df = gap[gap['country'].isin(['India', 'Brazil', 'Germany'])]
sns.relplot(data=temp_df, x='year', y='lifeExp', kind='line', hue='country',style='continent')
```

```python
#facet plot will only with figure level functions 
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',row='day',col='sex')
```

```python
sns.relplot(data=gap,x='lifeExp',y='gdpPercap',kind='scatter',col='year',col_wrap=2,hue='country')
```

```python
#distribution plots
#histograph  fig level function displot
#axes level function - histplot,kdeplot,rugplot

sns.histplot(data=tips,x='day',hue='sex',element='step') #used step step so we can se behind value
```

```python
sns.displot(data=tips,x='total_bill',kind='hist')
```

```python
sns.displot(data=titanic,x='age',kind='hist',hue='survived',element='step',col='sex')
```

```python
sns.displot(data=tips,x='total_bill',kind='kde',fill=True,hue='sex',height=10,aspect=2) #here the graph tells how much total_bill our probablity can be 
```

```python
#rugplot
sns.displot(data=tips,x='total_bill',kind='kde')
sns.rugplot(data=tips,x='total_bill')
```

```python
#baivariate histogram
sns.histplot(data=tips,x='total_bill',y='tip')
```

```python
sns.displot(data=tips,kind='kde',x='total_bill',y='tip')
```

```python
#matrix plots 
# heatmap
temp_df = gap.pivot(index='country',columns='year',values='lifeExp')
import matplotlib.pyplot as plt
plt.figure(figsize=(15,15))
sns.heatmap(temp_df)
```

```python
#annot
temp = gap[gap['continent'] == 'Europe'].pivot(index='country',columns='year',values='lifeExp')
plt.figure(figsize=(8,10))
sns.heatmap(temp,annot=True,linewidths=0.5,cmap='summer')
```

```python
#cluster map
iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```

```python
#categorial plots
sns.stripplot(data=tips,x='day',y='total_bill')
```

```python
#using catplot
sns.catplot(data=tips,x='day',y='total_bill',kind='strip',hue='sex')
```

```python
#sarm plot used for cheking data distribution
sns.catplot(data=tips,x='day',y='total_bill',kind='swarm')
```

```python
#categrorical distribtion plot
#boxplot
sns.boxplot(data=tips,x='day',y='total_bill')
```

```python
#we can only get info form only one column
sns.boxplot(data=tips,y='total_bill')
```

```python
#violineplot
sns.violinplot(data=tips,x='day',y='total_bill')
```

```python
#catigroical estimation plot (estimator used for mean max std and other method)
import numpy as np
sns.barplot(data=tips,x='sex',y='total_bill',hue='smoker',estimator=np.max,errorbar=None)
```

```python
#point plot it connets top of the both graph with a line
sns.pointplot(data=tips,x='sex',y='total_bill',hue='smoker',estimator=np.max,errorbar=None)
```

```python
#count plots used for counting the calues on x axis
sns.countplot(data=tips,x='sex',hue='day')
```

```python
sns.catplot(data=tips,x='sex',y='total_bill',col='smoker',kind='box',row='time')
```
