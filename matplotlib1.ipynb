```python
import matplotlib.pyplot as plt

x = [1,2,3,5,9]
y=[45,7,5,65,14]

plt.plot(x,y)
plt.show()
```

```python
x = ['mon','tues','wed','thur','fri'] #x-axis
y = [100,500,450,900,250] #y-axis

plt.plot(x,y)

plt.title('Bakery sales this week')

plt.xlabel('day of the week')
plt.ylabel('sales per day')

plt.show()
```

```python
from matplotlib.lines import lineStyles
#line chart
#plt.plot(x,y,color,line-style,line-width,marker-symbol,labeling) optional
months = [1,2,3,4,5]
sales = [1000,5000,6000,4000,3000]

plt.plot(months,sales,color="blue",linestyle="--",linewidth=2,marker='o',label="2025 sales data")
plt.xlabel("months")
plt.ylabel("sales for each months")
plt.title("monthly sales data report")
plt.legend(loc="upper left")#we add location to lower right
plt.grid(color="gray",linestyle=":",linewidth=1)
# plt.xlim(1,5)
# plt.ylim(0,5000)
plt.xticks([1,2,3,4,5],['M1','M2','M3','M4','M5'])
plt.show()
```

```python

#bar chart is used for comapring value with each other
#pi chart is used for cheaking the propotions
#histogram is used for cheakking stadard deviation so to chek how the data is spread out
# plt.bar(x,height,color,width,label) syntax

products = ['A','B','C','D','E']
sales = [1000,2000,6000,5000,4000]

plt.bar(products,sales,color='orange',label='sales 2025') #we can see data in horizontliy using barh
plt.xlabel('prodcuts')
plt.ylabel('sales')
plt.title('product sales comparison')
plt.legend()
plt.show()
```

```python
#pi chart syntax
#plt.pie(values,label,colors,autopercentage='%1.1f%%')

regions = ['north','south','East','west']
revenu = [3000,1000,5600,4000]

plt.pie(revenu,labels=regions,autopct='%1.1f%%',colors=['gold','skyblue','lightgreen','coral'])
plt.title('revenue contribution by region')
plt.show()
```

```python
#histogram syntax plt.hist(data,bins=number_of_bins,color,edgecolor)

scores = [45,89,47,65,12,35,47,58,56,48,45,65,74,48,85,93,18,45,65]

plt.hist(scores,bins=7,color='purple',edgecolor='black',label='student scores')
plt.xlabel('Score range')
plt.ylabel('Number of student')
plt.title('Score Distribution of student')
plt.legend()
plt.show()
```

```python
#scatter plot syntax used for corelation beteween two data
#plt.scatter(x,y,color,marker,label)
houres_studies = [1,2,3,4,5,6,7,8]
marks = [56,41,55,60,77,80,91,92]

plt.scatter(houres_studies,marks,color='green',marker='o',label='Class A')
plt.scatter([1,2,3,4,5,6,7,8],[45,74,85,75,90,91,95,99],color='red',marker='o',label='Class B')
plt.xlabel('Hours studied')
plt.ylabel('Exam Score')
plt.title('Relationship between two classes')
plt.legend()
plt.grid(True)
plt.show()
```

```python
#subplot functions
#what is subplots comparing charts
#plt.subplot(nrows,ncolumns,index)

x = [1,2,3,4]
y = [50,40,60,10]
plt.subplot(1,3,1) #1st row , 2nd column and 1st position
plt.plot(x,y)
plt.title('line chart')

plt.subplot(1,3,2)  #1st row , 2nd column and 2nd position
plt.bar(x,y)
plt.title('bar chart')

plt.subplot(2,3,1)
plt.hist(x,bins=5,color='purple',edgecolor='black',label='student scores')
plt.title('pie chart')

plt.tight_layout()
plt.show()
```

```python
fig,ax = plt.subplots(1,2,figsize=(10,5))
x = [1,2,3,4]
y = [50,40,60,10]

ax[0].plot(x,y, color='blue')
ax[0].set_title("line plot")

ax[1].bar(x,y,color='green')
ax[1].set_title("bar chart")

fig.subtitle('comparison of line and bar chart')

plt.tight_layout()
plt.show()
```

```python
#how to save a figure

x = [1,2,3,4]
y = [10,15,50,36]

plt.plot(x,y,color="blue",marker='o')
plt.title('simple line plot ')
plt.xlabel('X axis')
plt.ylabel('Y axis')

plt.savefig('line_plot.png',dpi=300,bbox_inches='tight')
plt.show()
```
