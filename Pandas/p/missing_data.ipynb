```python
#we are going to see how to handle missing data
#NAN(for a number that is not there) and none(for a object that is not there)
import pandas as pd
data = {
    "name": [
        "Rahul",None , "Amit", "Neha", "Karan",
        "Sneha", "Arjun", "Pooja", "Vikas", "Anjali",
        "Rohan", "Meera", "Yash", "Kavya", "Jay"
    ],

    "age": [
        25, None, 32, 24, 30,
        27, 35, 29, 26, 31,
        23, 34, 28, 22, 33
    ],

    "salary": [
        30000, None, 60000, 28000, 55000,
        40000, 75000, 48000, 35000, 65000,
        25000, 70000, 42000, 27000, 62000
    ],

    "performance_score": [
        78, None, 85, 70, 88,
        95, 82, 76, 68, 90,
        72, 89, 84, 96, 80
    ]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.isnull().sum()) #checks weather the value is  none or nan and sum will show how many values are missing
```

```python
#handilng the missing value first we are going to removing if the values are not affecting the current motive
# df.dropna(axis=0,inplace=True) #axis 0 means removing the sepefic rows and axis 1 is removing column
# print(df)

df['age'].fillna(df['age'].mean(),inplace=True) #here we are filling the empty value in the specifc column with putting mean function there so the value will be mean 
df['salary'].fillna(df['salary'].mean(),inplace=True)

print(df)

```

```python
#interpolation example , liner,ploynomial,time
#liner interploation
df["performance_score"].interpolate(method="linear" ,axis=0,inplace=True)
df["salary"].interpolate(method="linear" ,axis=0,inplace=True)
df["age"].interpolate(method="linear")
# z = df["performance_score"].interpolate( #we use this when we need values higher 
#     method="polynomial", order=2, axis=0, inplace=False
```

```python
#shorting and aggrigation used for managing and getting arranged data
print(df.sort_values(by=["performance_score","salary"],ascending=[True,True]))
```

```python
#grouping of data
groupd = df.groupby(["age" ,"name"])["salary"].sum()
print(groupd)
```

```python
#merging and joining
# Merging and joining
df1 = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Neha", "Karan"],
    "age": [25, 28, None, 24, 30],
    "salary": [30000, None, 50000, 28000, 55000],
    "performance_score": [70, 75, None, 90, 100]
})

df2 = pd.DataFrame({
    "name": ["Rahul", "Priya", "Amit", "Neha", "Karan"],
    "age": [25, 28, None, 24, 30],
    "product": ["Laptop", "Mouse", "Keyboard", "Monitor", "Headphones"],
    "price": [50000, 800, None, 12000, 2500],
    "units_sold": [10, None, 25, 15, 30],
    "rating": [4.5, 4.0, 3.5, None, 4.8]
})

vf = pd.merge(df1, df2, on="name", how="outer")
print(vf)
```

```python
#conncate the dataframes
k = pd.concat([df1,df2],axis=1,ignore_index=True)
# k = k.drop_duplicates(subset=["name"])
print(k)
```
