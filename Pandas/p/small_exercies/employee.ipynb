```python
#importing library
import pandas as pd
import numpy as np 
```

```python
df = pd.read_csv(r"C:\Users\dgajj\OneDrive\Desktop\intern\softvan techology(python)\Pandas\p\small_exercies\Employee.csv")

print(df.head())
```

```python
#cheking the missing value
print(df.isnull().sum()) 
```

```python
#filling the salaryby getting the mean and also
df['Salary (INR)'].fillna(df['Salary (INR)'].mean(),inplace=True)
print(df.head())
df['Performance Rating'].fillna(df['Performance Rating'].median(),inplace=True)
print(df.head())
```

```python
df.replace([np.inf, -np.inf], np.nan, inplace=True)
df.fillna(df.mean(numeric_only=True), inplace=True)
print(df)
```

```python
print(df[df.duplicated(keep=False)]) #checking if there is nay duplicated columns
df.drop_duplicates(inplace=True)

```

```python
#replacing negetive value with positive values
df['Salary (INR)'] = np.where(df['Salary (INR)'] < 0, df['Salary (INR)'].mean(),df['Salary (INR)'])
```

```python
salary_mean = df['Salary (INR)'].mean()
salary_std = df['Salary (INR)'].std()
lower_bound = salary_mean - (3* salary_std)
upper_bound = salary_mean + (3* salary_std)

#remove rows where salary is to high and too low
df= df[(df['Salary (INR)'] >= lower_bound ) & (df['Salary (INR)'] <= upper_bound ) ]
```

```python
df.to_csv('cleaned_data.csv',index=False)
```
