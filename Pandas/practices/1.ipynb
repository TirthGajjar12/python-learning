```python
import pandas as pd
import numpy as np

data = { 
    'ticket_id': ['T-101', 'T-102', 'T-103', 'T-104', 'T-105', 'T-106', 'T-107', 'T-108'],
    'asset_type': ['Laptop', 'Monitor', 'Laptop', 'Server', 'Keyboard', 'Laptop', 'Router', 'Monitor'],
    'department': ['HR', 'Engineering', 'Marketing', 'Engineering', 'HR', 'Engineering', 'IT', 'Marketing'],
    'resolution_time_hrs': [4.5, 1.2, np.nan, 36.0, 0.5, 12.5, np.nan, 2.0],
    'hardware_cost_inr': [65000, 15000, 72000, 250000, 1500, 85000, 12000, np.nan],
    'satisfaction_score': [4, 5, 3, 2, 5, np.nan, 4, 4]
}

df = pd.DataFrame(data)
print(df) 
```

```python
# print(df.tail(4))
print(df.isnull().sum())
```

```python

#filling the missing value of the hardware_cost_inr with the median cost of the existing hardware
df['hardware_cost_inr'] = df['hardware_cost_inr'].fillna(df['hardware_cost_inr'].median())
print(df)
```

```python
print(df['resolution_time_hrs'].isnull())
```

```python
df = df.dropna(subset=['resolution_time_hrs'])  
print(df)
```

```python
df.loc[5,'satisfaction_score'] = 3
print(df)
```

```python

grouped = df.groupby(['department'])['hardware_cost_inr'].sum()
print(grouped)

co = df.groupby(['asset_type'])['ticket_id'].count()
print(co)
```

```python
arr = np.array(df['resolution_time_hrs'])
print(arr)
```

```python
std = arr.std()
print(std)

mean = arr.mean()
print(mean)

time_hrs = arr.astype(int)
print(time_hrs)
```

```python
ticket_ids = df.loc[(df['resolution_time_hrs'] > 10 ) & (df['satisfaction_score'] < 4),'ticket_id']
print(ticket_ids)
```
