**Example**

```python
def longest_increasing_run(numbers):
    # YOUR LOGIC HERE
    longest = current = 1

    for i in range(1,len(numbers)):
        if numbers[i] == numbers[i-1]+1:
            current +=1
        else:
            current = 1

        longest = max(longest,current)
    return longest


# Test input
numbers = [1, 2, 3, 2, 4, 5, 6, 7, 1]
result = longest_increasing_run(numbers)
print(result)
```
