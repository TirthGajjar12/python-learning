```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score

#true answers (what actally happend)
y_true = [1,0,1,1,0,1,0]

#model predication (what it guessed)
y_pred = [1,0,1,0,0,1,1]

#evalution

print(accuracy_score(y_true,y_pred)) #accuracy socre
print(precision_score(y_true,y_pred))
print(recall_score(y_true,y_pred))
print(f1_score(y_true,y_pred))
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error
import numpy as np

real_score = [90,60,80,100]

predicated_scores = [85,70,70,95]

mae = mean_absolute_error(real_score,predicated_scores)
mse = mean_squared_error(real_score,predicated_scores)
rmse = np.sqrt(mse)

print(mae,mse,rmse)
```
