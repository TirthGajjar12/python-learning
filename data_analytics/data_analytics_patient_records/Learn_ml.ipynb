```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Name': ['Pavan','Kapil','Lalita','Ishan','Yashvi'],
    'Age' : [25,None,44,23,None],
    'Salary' : [50000,60000,70000,None,None],
    'Gender':['Male','Male','Female','Male','Female']
}
```

```python
df = pd.DataFrame(data)
df
```

```python
df.isnull().sum()
```

```python
df = df.fillna(df[['Age','Salary']].mean())
df
```

```python
#before gigving data to our model we have to preprocces the data to make sur ethere no other data that remaning 
from sklearn.preprocessing import LabelEncoder

#label encoding
le = LabelEncoder()
df['Gender_Encoded'] = le.fit_transform(df['Gender'])  #converting values to label encoding so this values are encoded in numaric so can be proccesed by our model
df
```

```python

df['City']  = ['Indor','Delhi','Mumbai','Goa','Ahmedabad']
df
```

```python
#onehot encoding 
df = pd.get_dummies(df,columns=['City'])
df
```

```python
city_col = ['City_Ahmedabad','City_Delhi','City_Goa','City_Indor','City_Mumbai']

df[city_col] = df[city_col].astype(int) #converted hot coded true flase into label encoding

df['City_Label'] = df[city_col].idxmax(axis=1)

df['City_Label'] = le.fit_transform(df['City_Label'])

df
```

```python
#feature scaling 
# 1 standard scaler
from sklearn.preprocessing import StandardScaler,MinMaxScaler

scaler = StandardScaler()
x_scaler = scaler.fit_transform()

scaler = MinMaxScaler()
x_scaler = scaler.fit_transform()

```

```python
from sklearn.model_selection import train_test_split

data = {
    'StudyHours' : [1,2,3,4,5],
    'TestScore' : [40,50,60,70,80]
}

df = pd.DataFrame(data)


#using standardsclaer

standard_scaler = StandardScaler()
standard_scaled =   standard_scaler.fit_transform(df)
pd.DataFrame(standard_scaled,columns=['StudyHours','TestScore'])

#minmax Scaler

minmax_scaler = MinMaxScaler()
minmax_scaled = minmax_scaler.fit_transform(df)
pd.DataFrame(minmax_scaled,columns=['StudyHours','TestScore'])
```

```python
#strain_test_split usage
x = df[['StudyHours']]
y = df[['TestScore']]

X_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
print(X_train)
print(x_test)
print(y_train)
print(y_test)
```
