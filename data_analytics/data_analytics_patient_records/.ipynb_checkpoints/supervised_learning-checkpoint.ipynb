```python
#linear regression 
#finds a pattern in old data 
#straight line 
#line Y= mx+b

#its not about line its about story
#accuarcy is not always the goal 
from sklearn.linear_model import LinearRegression

x= [[1],   #we need to write this x way because it expactes 2d array
    [2],
    [3],
    [4],
    [5]
   ]

y = [40,50,65,75,90] #if we are expeacting 1d array in output then we can write this

model = LinearRegression()
model.fit(x,y)

hours = float(input("Enter how many hours you studied = "))

predicted_marks = model.predict([[hours]])

print(predicted_marks)
```

```python
#classification means claasiying new things into group 
#label or categoires
#logistic regression 
#outcome will be in binary 0 or 1 
from sklearn.linear_model import LogisticRegression

x= [[1],   #we need to write this x way because it expactes 2d array
    [2],
    [3],
    [4],
    [5]
   ]
#here x means hours studied and y means if 0 failed and 1 means passed 
y = [0,0,1,1,1]

model = LogisticRegression()
model.fit(x,y)

hours = float(input("Enter how many hours you studied = "))

predicted_marks = model.predict([[hours]])[0] #Because it always returns an array, index [0] grabs the first (and only) item out of that array

def result_F():
    if predicted_marks == 1:
        return f"you have passed exam {predicted_marks}"
    else:
        return f"you have failed exam {predicted_marks}"

result_F()
```

```python
#knn gets colsest to the values means that whenever we use that if the mejority values are saying yes and other are saying no then it will go for yes 
#knn is slower for big numbers of data 

from sklearn.neighbors import KNeighborsClassifier

x = [
    [180,7],
    [200,7.5],
    [250,8],
    [300,8.5],
    [330,9],
    [360,9.5]
]

# 0 = Apple and 1 = Orange
y = [0,0,0,1,1,1]

model = KNeighborsClassifier()
model.fit(x,y)

weight = float(input('Enter the weight in grams: '))
height = float(input('Enter the weight in cms: '))

predicted_fruite = model.predict([[weight,height]])[0]

def predict_fruits():
    if predicted_fruite == 1:
        return f"the fruits diamantions that you have given looks like Orange"
    else:
        return f"the fruits diamantions that you have given looks like Apple"

predict_fruits()
```

```python
#decision tree 
from sklearn.tree import DecisionTreeClassifier

x = [
    [7,2], #apple
    [8,3], #orange
    [9,8], #Orange
    [10,9] #Orange

]

y = [0,0,1,1] # 0 = Apple and 1 = Orange

model = DecisionTreeClassifier()
model.fit(x,y)
size = float(input('Enter the fruits size: '))
shade = float(input('Enter the fruits shade: '))

results  = model.predict([[size,shade]])[0]

def get_result():
    if results == 1:
        return "By the size and shade you have given me it feels like it is Orange"
    else:
        return "By the size and shade you have given me it feels like it is Apple"

get_result()
```
