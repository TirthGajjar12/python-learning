```python
#how long patients satys in each department 
department_los = df.groupby("department").agg(
    average_stay=("length_of_stay_days", "mean"),
    median_stay=("length_of_stay_days", "median"),
    maximum_stay=("length_of_stay_days", "max"),
    patient_count=("patient_id", "count")
)

department_los = department_los.sort_values(
    by="average_stay",
    ascending=False
)

print(department_los.round(2))
```

```python
#Calculating the 90th percentile for each department
department_los["p90_stay"] = (
    df.groupby("department")["length_of_stay_days"]
    .agg(lambda x: np.percentile(x, 90))
)

department_los = department_los.sort_values(
    by="p90_stay",
    ascending=False
)

print(
    department_los[
        ["average_stay", "median_stay", "p90_stay", "maximum_stay"]
    ].round(2)
)
```

```python
department_los["std_stay"] = (
    df.groupby("department")["length_of_stay_days"].std()
)

print(
    department_los[
        ["average_stay", "median_stay", "p90_stay", "std_stay"]
    ].round(2)
)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 2, figsize=(18, 8))

plot_data = department_los.reset_index()

# Graph 1: Average vs Median
comparison_data = plot_data.melt(
    id_vars="department",
    value_vars=["average_stay", "median_stay"],
    var_name="stay_type",
    value_name="days"
)

sns.barplot(
    data=comparison_data,
    x="days",
    y="department",
    hue="stay_type",
    ax=axes[0]
)

axes[0].set_title("Average vs Median Length of Stay")
axes[0].set_xlabel("Days")
axes[0].set_ylabel("Department")

# Graph 2: Average vs P90
sns.scatterplot(
    data=plot_data,
    x="average_stay",
    y="p90_stay",
    size="patient_count",
    sizes=(40, 250),
    hue="department",
    ax=axes[1]
)

axes[1].set_title("Average Stay vs 90th Percentile")
axes[1].set_xlabel("Average Stay (Days)")
axes[1].set_ylabel("P90 Stay (Days)")

plt.tight_layout()
plt.show()
```

```python
#Insurance analytics
insurance_analysis = df.groupby("insurance_provider").agg(
    patient_count=("patient_id", "count"),
    total_billing=("total_bill_inr", "sum"),
    insurance_covered=("insurance_covered_inr", "sum"),
    patient_paid=("patient_paid_inr", "sum"),
    outstanding_balance=("outstanding_balance_inr", "sum")
)

insurance_analysis["coverage_percentage"] = (
    insurance_analysis["insurance_covered"]
    / insurance_analysis["total_billing"]
) * 100

insurance_analysis = insurance_analysis.sort_values(
    by="total_billing",
    ascending=False
)

print(insurance_analysis.round(2))
```

```python
insurance_analysis["avg_bill_per_patient"] = (
    insurance_analysis["total_billing"]
    / insurance_analysis["patient_count"]
)

insurance_analysis["avg_covered_per_patient"] = (
    insurance_analysis["insurance_covered"]
    / insurance_analysis["patient_count"]
)

insurance_analysis["avg_outstanding_per_patient"] = (
    insurance_analysis["outstanding_balance"]
    / insurance_analysis["patient_count"]
)

print(
    insurance_analysis[
        [
            "patient_count",
            "avg_bill_per_patient",
            "avg_covered_per_patient",
            "avg_outstanding_per_patient"
        ]
    ].round(2)
)
```

```python
insurance_plot = insurance_analysis.reset_index()

coverage_comparison = insurance_plot.melt(
    id_vars="insurance_provider",
    value_vars=[
        "avg_bill_per_patient",
        "avg_covered_per_patient"
    ],
    var_name="amount_type",
    value_name="amount"
)

# Create two charts in one figure
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# Chart 1: Average bill vs insurance coverage
sns.barplot(
    data=coverage_comparison,
    x="amount",
    y="insurance_provider",
    hue="amount_type",
    ax=axes[0]
)

axes[0].set_title("Average Bill vs Insurance Coverage per Patient")
axes[0].set_xlabel("Amount (INR)")
axes[0].set_ylabel("Provider / Category")
axes[0].legend(title="Amount Type")

# Chart 2: Average outstanding balance
sns.barplot(
    data=insurance_plot.sort_values(
        "avg_outstanding_per_patient",
        ascending=False
    ),
    x="avg_outstanding_per_patient",
    y="insurance_provider",
    color="steelblue",
    ax=axes[1]
)

axes[1].set_title("Average Outstanding Balance per Patient")
axes[1].set_xlabel("Outstanding Amount (INR)")
axes[1].set_ylabel("Provider / Category")

plt.tight_layout()
plt.show()
```

```python
payment_distribution = pd.crosstab(
    df["insurance_provider"],
    df["payment_status"]
)

print(payment_distribution)
```

```python
payment_percentage = pd.crosstab(
    df["insurance_provider"],
    df["payment_status"],
    normalize="index"
) * 100

print(payment_percentage.round(2))
```

```python
fig, ax = plt.subplots(figsize=(13, 7))

payment_percentage.plot(
    kind="barh",
    stacked=True,
    ax=ax,
    figsize=(13, 7)
)

ax.set_title("Payment Status Distribution by Insurance Provider")
ax.set_xlabel("Percentage of Patients (%)")
ax.set_ylabel("Provider / Category")
ax.set_xlim(0, 100)

ax.legend(
    title="Payment Status",
    bbox_to_anchor=(1.02, 1),
    loc="upper left"
)

plt.tight_layout()
plt.show()
```

```python
from scipy.stats import chi2_contingency

chi2, p_value, degrees_of_freedom, expected = chi2_contingency(
    payment_distribution
)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", degrees_of_freedom)
```

```python
diagnosis_frequency = (
    df["primary_diagnosis"]
    .value_counts()
    .reset_index()
)

diagnosis_frequency.columns = [
    "diagnosis",
    "patient_count"
]

diagnosis_frequency["percentage"] = (
    diagnosis_frequency["patient_count"]
    / len(df)
) * 100

print(diagnosis_frequency.to_string(index=False))
```

```python
fig, ax = plt.subplots(figsize=(12, 8))

top_diagnoses = diagnosis_frequency.head(15)

sns.barplot(
    data=top_diagnoses,
    x="patient_count",
    y="diagnosis",
    color="steelblue",
    ax=ax
)

ax.set_title("Top 15 Most Frequent Primary Diagnoses")
ax.set_xlabel("Number of Patients")
ax.set_ylabel("Primary Diagnosis")

plt.tight_layout()
plt.show()
```

```python
age_diagnosis = df.groupby("primary_diagnosis").agg(
    patient_count=("patient_id", "count"),
    average_age=("age", "mean"),
    median_age=("age", "median"),
    std_age=("age", "std")
)

age_diagnosis = age_diagnosis.sort_values(
    by="patient_count",
    ascending=False
)

print(age_diagnosis.head(15).round(2))
```

```python
top_10_diagnoses = diagnosis_frequency.head(10)["diagnosis"]

age_plot_data = df[
    df["primary_diagnosis"].isin(top_10_diagnoses)
].copy()

fig, axes = plt.subplots(1, 2, figsize=(20, 8))

# Chart 1: Average age
sns.barplot(
    data=age_diagnosis.reset_index().head(10),
    x="average_age",
    y="primary_diagnosis",
    color="steelblue",
    ax=axes[0]
)

axes[0].set_title("Average Patient Age by Diagnosis")
axes[0].set_xlabel("Average Age (Years)")
axes[0].set_ylabel("Diagnosis")

# Chart 2: Age distribution
sns.boxplot(
    data=age_plot_data,
    x="age",
    y="primary_diagnosis",
    order=top_10_diagnoses.tolist(),
    ax=axes[1]
)

axes[1].set_title("Age Distribution by Diagnosis")
axes[1].set_xlabel("Age (Years)")
axes[1].set_ylabel("Diagnosis")

plt.tight_layout()
plt.show()
```
